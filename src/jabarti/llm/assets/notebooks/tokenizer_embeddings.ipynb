# Tokenizer

```python
from jabarti.llm import tokenizer

tok = tokenizer.Tokenizer.from_file()
print(f"vocab_size:{tok.vocab_size}")
```

```python
sentence = "ابوبكر     سليمان"

norm_text = tok.normalize(sentence)

print(norm_text)
```

```python
print(tok._tokenizer.normalizer)
print(tok._tokenizer.to_str()[:1500])  # look at the "normalizer" section
```

```python
for t in ["مُحَمَّد", "محـــمد", "أحمد إبراهيم آمنة", "على", "a    b"]:
    print(repr(t), "->", repr(tok.normalize(t)))
```

```python
encoded = tok.encode(sentence)
print(f"Encoded:{encoded}")
```

```python
tokenize = tok.tokenize(sentence)
print(f"Tokenizes:{tokenize}")
```

```python
dec = tok.decode([7904, 8106, 8144, 12149])
print(f"decode:{dec}")
```

# Embeddings

```python
from jabarti.llm.config import ModelConfig
from jabarti.llm.model.embeddings import InputEmbedding
import torch
```

```python


my_config = ModelConfig()
embed = InputEmbedding(config=my_config)


sentence_1 = "ابوبكر سليمان    Abu Bakr Soliman "
sentence_2 = "ابوبكر سليمان    Abu Bakr Soliman "
```

```python
encoded_1 = tok.encode(sentence_1)
encoded_2 = tok.encode(sentence_2)

input_ids = torch.tensor([
    encoded_1,
    encoded_2
])
```

```python
input_vectors = embed(input_ids)
input_vectors.shape
```
